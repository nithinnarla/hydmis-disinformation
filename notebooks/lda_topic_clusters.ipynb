- Stage 2 GPT-4 semantic verification needed for veracity discrimination
- LIAR2: integer labels 0=pants-fire to 5=true; TruthSeeker: Agree/Disagree/NO MAJORITY; FakeNewsNet: 0=real/1=fake

## 10. Key Findings

**Partial topic-veracity alignment:** LDA captures content domains (health misinformation, political claims, social media narratives) but not veracity directly. No topic is purely disinformation or purely credible.

**Cross-dataset consistency:** Topic structure holds across LIAR2, TruthSeeker, and FakeNewsNet, validating that English LDA discovers generalizable content clusters.

**Stage 2 implication:** Unsupervised topic modeling is insufficient for veracity discrimination. GPT-4 semantic verification (Stage 2) is needed to move from content clusters to credibility assessment.

**Label encoding confirmed:**
- LIAR2: integer 0=pants-fire → 5=true
- TruthSeeker: Agree/Mostly Agree/NO MAJORITY/Mostly Disagree/Disagree
- FakeNewsNet: 0=real, 1=fake
